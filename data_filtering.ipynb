```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("/content/grade-records.csv")
```

```python
df.head()
```

```python

df.columns = df.columns.str.replace(r'\s+', ' ', regex=True).str.strip()

averages = pd.to_numeric(df['Student Average'].str.replace('%', '', regex=False), errors='coerce')
data = averages.dropna().to_numpy()

print("NumPy array of student averages:")
print(data)
print("\nNumber of numerical values:", len(data))
```

```python
# Use the cleaned numerical data array we created earlier
student_averages = data

print("Student averages:", student_averages)
print("Number of valid averages:", len(student_averages))
```

```python
threshold = 50

above = data[data > threshold]

print("Values above", threshold, ":")
print(above)
```

```python
threshold = 50

below = data[data < threshold]

print("Values below", threshold, ":")
print(below)
```

```python
threshold = 50

below = data[data < threshold]

print("Values below", threshold, ":")
print(below)
```

```python
thresholds = [40, 60, 70, 80]

for threshold in thresholds:
    print(f"\nThreshold: {threshold}")
    print("Above:", data[data > threshold])
    print("Below:", data[data < threshold])
    print("Equal:", data[data == threshold])
```
